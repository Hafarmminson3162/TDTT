Bài 6: Nhập vào 3 số a, b, c. Kiểm tra xem đây có phải là cạnh của 1 tam giác không, nếu đúng thì tính diện tích của tam giác đó, in ra với 1 chữ số phần thập phân, nếu nó không phải 3 cạnh của tam giác thì in ra "Khong phai 3 canh cua tam giac".

```python
a, b, c=map(int, input("Nhap a,b,c").split())
c, b, a = sorted([a, b, c])
if c > 0 and (b + c > a):
    if (c**2+b**2==a**2):
        print("Dien tich tam giac vuong: ", b*c/2)
    else:
        print("Day la 3 canh cua tam giac")
else:
    print("Khong phai 3 canh cua tam giac")
```

Bài 8: Nhập vào họ tên chủ hộ và chỉ số điện kế tháng trước và chỉ số điện kế tháng này, tính tiền điện tháng này cho hộ biết:

- Từ 0 – 50kWh: 1.984 đồng/kWh

- Bậc 2: Từ 51 – 100 kWh: 2.050 đồng/kWh

- Bậc 3: Từ 101 – 200 kWh: 2.380 đồng/kWh

- Bậc 4: Từ 201 – 300 kWh: 2.998 đồng/kWh

- Bậc 5: Từ 301 - 400 kWh: 3.350 đồng/kWh

- Bậc 6: Từ 401 trở đi kWh: 3.460 đồng/kWh

Tính thêm tiền VAT 8%. Chỉ lấy đến đơn vị đồng.



Ví dụ Input / Output:

+ Input 1:

  Ten chu ho: Pham Xuan Toai

  Chi so thang truoc: 1020

  Chi so thang nay: 1070

  Output 1:

  Ho va ten: Pham Xuan Toai

  Tien phai tra la: 107136



+ Input 2:

  Ten chu ho: Nguyen Xuan Long

  Chi so thang truoc: 2097

  Chi so thang nay: 2278

  Output 2:

  Ho va ten: Nguyen Xuan Long

  Tien phai tra la: 426038



+ Input 3:

  Ten chu ho: Nguyen Xuan Hai

  Chi so thang truoc: 2000

  Chi so thang nay: 2586

  Output 3:

  Ho va ten: Nguyen Xuan Hai

  Tien phai tra la: 1855505

```python
ten=str(input("Ten chu ho: "))
truoc=int(input("Chi so thang truoc: "))
sau=int(input("Chi so thang nay: "))
chiso=sau-truoc
if (chiso<=50):
     tien=chiso*1984
elif (chiso<=100):
     tien=(chiso-50)*2050 + 50*1984
elif (chiso<=200):
     tien=(chiso-100)*2380 + 50*2050 + 50*1984
elif (chiso<=300):
     tien=(chiso-200)*2998 +100*2380 + 50*2050 + 50*1984
elif (chiso<=400):
     tien=(chiso-300)*3350 +100*2998 +100*2380 + 50*2050 + 50*1984
else:
     tien=(chiso-400)*3460+100*3350 +100*2998 +100*2380 + 50*2050 + 50*1984
tien=float(tien*1.08)
mon=int(tien)
if (tien-mon>=0.5):
     mon+=1
print("Ho va ten: ", ten)
print("Tien phai tra la: ", mon)
```
