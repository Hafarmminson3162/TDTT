Bài 3 : Tính tam giác: Nhâp vào ba số a,b,c bất kì. Kiểmtra xem nó có thể làđộ dài ba cạnh hay không, nếu không thì in ra màn hình 'Khong la ba canh cua mot tam giac'. Ngược lại, kiểm tra xem nó là tam giác gì: Đều, Cân, Vuông, hay Tam giác thường. Tính diện tích, chu vi của tam giác và in ra mànhình.

```python
import math
a,b,c=map(float,input().split())
a,b,c=sorted([a,b,c])
if (a+b>c) and a>0:
    if (a==b==c): print("Tam giac deu")
    elif (a==b): print("Tam giac can")
    elif (c**2==a**2+b**2): print("Tam giac vuong")
    else: print("Tam giac thuong")

    print("Chu vi la",a+b+c)
    p=(a+b+c)/2
    s=math.sqrt(p*(p-a)*(p-b)*(p-c))
    print(f"Dien tich la {s:.2f}")
else:
    print("Khong la ba canh cua mot tam giac")
```

Bài 5: Viết chương trình giải phương trình bậc 2 (ax2 + bx + c = 0)

```python
import math
a,b,c=map(float,input().split())
if a==0:
    if b==0:
        if c==0:
            print("Vo so nghiem")
        else:
            print("Vo nghiem")
    else:
        x=-c/b
        print(f"Có 1 nghiệm: {x}")
else:
    delta =b**2-4*a*c
    if delta<0:
        print("Vo nghiem")
    elif delta==0:
        x=-b/(2*a)
        print(f"Co nghiem kep: {x}")
    else:
        x1=(-b+math.sqrt(delta))/(2*a)
        x2=(-b-math.sqrt(delta))/(2*a)
        print("Co 2 nghiem:")
        print(f"x1 = {x1}")
        print(f"x2 = {x2}")
```

Bài 8: Giải hệ phương trình tuyến tính:
ax  by  m
cx  dy  n

```python
a,b,m=map(float,input().split())
c,d,n=map(float,input().split())
if ((a*d-c*b)!=0):
    x=(m*d-n*b)/(a*d-c*b)
    y=(a*n-c*m)/(a*d-c*b)
    print(f"x = {x:.2f}; y = {y:.2f}")
elif (m*d-n*b==0):
    print ("Vo so nghiem")
else: print("Vo nghiem")
```

Bài 10: Nhập vào tâm và bán kính của một đường tròn. Sau đó nhập vàomộtđiểm A(x, y) bất kì và kiểm tra xem nó có thuộc đường tròn hay không?

```python
xi, yi, r=map(int,input().split())
x,y=map(int,input().split())
kc2=(x-xi)**2+(y-yi)**2
if (kc2==r**2): print(f"Diem ({x}, {y}) thuoc duong tron")
elif (kc2<r**2): print(f"Diem ({x}, {y}) thuoc hinh tron")
else : print(f"Diem ({x}, {y}) khong thuoc hinh tron")
```
